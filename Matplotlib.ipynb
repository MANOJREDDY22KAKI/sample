# This is one of the most common "gotchas" when starting with Matplotlib!

The reason you see a blank image is that plt.show() actually clears the current figure once you close the interactive window.

The Logic Behind the Blank Image
Think of Matplotlib like a physical canvas:

plt.plot(): You draw a picture on the canvas.

plt.show(): You show the canvas to the audience. Once the audience is done looking (you close the window), Matplotlib "cleans" the canvas so it’s ready for the next drawing.

plt.savefig(): You try to take a photo of the canvas, but it’s already been wiped clean.

```python
x = [1 ,2, 3, 4]
dsa_marks = [90, 80, 85, 70]
ml_marks = [70, 75, 80, 60]
plt.plot(x,dsa_marks, label='dsa_marks', marker='o',linewidth=2,linestyle='-.',markersize=8)
plt.plot(x,ml_marks, label='ml_marks', marker='*',linewidth=2,linestyle=':',markersize=10)
plt.title("Student Marks vs Hours Studied",fontdict={'fontname':'Comic Sans MS', 'fontsize':20, 'color':'purple'})
plt.xlabel("No of hours studied")
plt.ylabel("Marks Obtained")
plt.xticks([1,2,3,4], ['1 hour', '2 hours', '3 hours', '4 hours'])
plt.legend(loc='lower left')
plt.savefig('student_marks.png')
plt.show()
```

```python
x = np.linspace(0, 10, 100)

fig , ax = plt.subplots(2,3)
ax[0,0].plot(x, np.sin(x))
ax[0,0].set_title('Sine Wave')
ax[0,0].set_xlabel('X-axis')
ax[0,0].set_ylabel('Y-axis')

ax[0,1].plot(x, np.cos(x))
ax[0,1].set_title('Cosine Wave')
ax[0,1].set_xlabel('X-axis')
ax[0,1].set_ylabel('Y-axis')

ax[0,2].plot(x, np.tan(x))
ax[0,2].set_title('Tangent Wave')
ax[0,2].set_xlabel('X-axis')
ax[0,2].set_ylabel('Y-axis')

ax[1,0].plot(x, np.sinh(x))
ax[1,0].set_title('Hyperbolic Sine')
ax[1,0].set_xlabel('X-axis')
ax[1,0].set_ylabel('Y-axis')

ax[1,1].plot(x, np.cosh(x))
ax[1,1].set_title('Hyperbolic Cosine')
ax[1,1].set_xlabel('X-axis')
ax[1,1].set_ylabel('Y-axis')    

ax[1,2].plot(x, np.tanh(x))
ax[1,2].set_title('Hyperbolic Tangent')
ax[1,2].set_xlabel('X-axis')
ax[1,2].set_ylabel('Y-axis')    

plt.tight_layout()
plt.show()  
```

# Pie Chart

1. autopct='%1.1f%%' (Automatic Percentage)
This parameter tells Matplotlib to calculate the percentage for each slice automatically and display it as text. The string inside is a String Formatting code:

%: Start of the formatting command.

1.1f: This means "display as a float with 1 digit after the decimal point."

If you used %1.2f, it would show 24.15%.

If you used %1.0f, it would show 24%.

%%: This is how you tell Python to print a literal "%" symbol at the end of the number.

2. startangle=90 (Rotation)
By default, Matplotlib starts drawing the first slice of the pie at the 3 o'clock position (0 degrees) and moves counter-clockwise.

startangle=90 rotates the start point to the 12 o'clock position.

This is purely aesthetic, but many people find it looks more professional or "standard" to have the first slice start at the top.

```python
import matplotlib.pyplot as plt

Courses = ['DSA', 'ML', 'DL', 'AI', 'CN']
enrollments = [120, 100, 90, 80, 110]
hatches = ['/', '\\', '|', '-', '+']
colors = ['red', 'blue', 'green', 'orange', 'purple']

# FIX: Unpack the return values. 'patches' contains the actual pie slices.
patches, texts, autotexts = plt.pie(enrollments, labels=Courses, autopct='%1.1f%%', startangle=90)

# Iterate through 'patches' to apply styles
for patch, hatch, color in zip(patches, hatches, colors):
    patch.set_hatch(hatch)
    patch.set_facecolor(color) # Use set_facecolor for pie slices
    patch.set_edgecolor('black')

plt.title('Course Enrollments')
# Removed xlabel/ylabel as they aren't used in pie charts
plt.show()
```

```python
Courses = ['DSA', 'ML', 'DL', 'AI', 'CN']
enrollments = [120, 100, 90, 80, 110]
explode = (0.2, 0, 0, 0, 0)  # explode first slice
colors = ['red', 'blue', 'green', 'orange', 'purple']
plt.pie(enrollments,labels=Courses, autopct='%.2f%%',explode=explode,colors=colors)
plt.title('Course Enrollments')
plt.legend(loc='lower right')
plt.show()
```

# 3D Bar

```python
ax = plt.axes(projection = '3d')
x = np.random.randint(1,10,5)
y = np.random.randint(1,10,5)
z = np.zeros(5)

dx = np.ones(5)
dy = np.ones(5)
dz = np.random.randint(1,10,5)

ax.set_title('3D Surface Plot')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.bar3d(x, y, z, dx, dy, dz, color='cyan', edgecolor='black', alpha=0.7)
plt.show()
```

```python
ax = plt.axes(projection='3d')

x = np.arange(-5, 5, 0.25)
y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

ax.set_title('3D Surface Plot')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
plt.show()
```

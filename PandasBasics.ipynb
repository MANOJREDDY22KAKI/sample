```python
df1 = pd.DataFrame({
    'id' : [1,2,3,4],
    'name' : ['Keerti', 'Amit', 'Educosys', 'YouTube']
})

df2 = pd.DataFrame({
    'id' : [3,4, 5, 6],
    'course' : ['GenAI', 'CPP', 'HLD', 'LLD']
})

inner_result = pd.merge(df1, df2, how='inner', on='id')
inner_result
```

```python
outer_merge = pd.merge(df1, df2,  how='outer', on='id')
outer_merge
```

```python
left_merge = pd.merge(df1, df2, how='left', on='id')
print(left_merge)
```

```python
df3 = pd.DataFrame({
    'id1' : [1,2,3,4],
    'name' : ['Keerti', 'Amit', 'Educosys', 'YouTube']
})

df4 = pd.DataFrame({
    'id2' : [3,4, 5, 6],
    'course' : ['GenAI', 'CPP', 'HLD', 'LLD']
})

left_merge_diff_cols = pd.merge(df3, df4, how='left', left_on='id1', right_on='id2')
left_merge_diff_cols
```

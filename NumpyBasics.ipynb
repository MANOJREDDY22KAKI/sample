axis = 0 means column wise elements will be sorted
axis = 1 means rows wise elements will be sorted

```python
arr = np.array([[14, 21, 13], [42, 15, 46], [27, 83, 92]])
arr_sorted_1 = np.sort(arr, axis=1)
arr_sorted_0 = np.sort(arr, axis=0)
#print(arr_sorted_1)
print("Original array:\n", arr)
print("Sorted array along rows:\n", arr_sorted_1)
print("Sorted array along columns:\n", arr_sorted_0)
```

# The numpy.argsort() function performs an indirect sort on a NumPy array, returning an array of indices that would put the original array's elements into ascending order. It does not modify the original array. 


```python
print("argsort along rows:\n", np.argsort(arr, axis=1))
print("argsort along columns:\n", np.argsort(arr, axis=0))
```

```python
print("original array:\n", arr)
print("lexsort along rows:\n", np.lexsort(arr.T))
print("lexsort along columns:\n", np.lexsort(arr))
print("searchsorted along rows:\n", np.searchsorted(arr[0], 15))
print("searchsorted along columns:\n", np.searchsorted(arr[:,0], 30))
print("partition along rows:\n", np.partition(arr, 1, axis=1))
print("partition along columns:\n", np.partition(arr, 1, axis=0))
```

```python
arr1 = np.array([3, 1, 2])
arr2 = np.array([6, 4, 5])
print("Original arrays:\n", arr1, arr2)
print("Concatenated array:\n", np.concatenate((arr1, arr2)))
print("Stacked array along axis 0:\n", np.stack((arr1, arr2), axis=0))
print("Stacked array along axis 1:\n", np.stack((arr1, arr2), axis=1))
```

```python
# reshape
arr = np.arange(12)
print("Original array:\n", arr)
reshaped_arr = arr.reshape((2, 6))
print("Reshaped array (2x6):\n", reshaped_arr)
```

```python
array = np.arange(8)
a1 = array[np.newaxis, :]
a2 = array[:, np.newaxis]
print("Original array shape:", array.shape)
print("After adding new axis at position 0:", a1.shape)
print("After adding new axis at position 1:", a2.shape)
print("Original array:\n", array)
print("After adding new axis at position 0:\n", a1)
print("After adding new axis at position 1:\n", a2)
```

```python
b1 = np.expand_dims(array, axis=0)
b2 = np.expand_dims(array, axis=1)
print("Using expand_dims - shape after adding new axis at position 0:", b1.shape)
print("Using expand_dims - shape after adding new axis at position 1:", b2.shape)
print("original array:\n", array)
print("Using expand_dims - after adding new axis at position 0:\n", b1)
print("Using expand_dims - after adding new axis at position 1:\n", b2)
```
